El Sr. y la Sra. H han planeado celebrar su 25º aniversario de bodas en Honeymooners' Paradise, un popular complejo turístico en la ciudad de Centerville. Contando hoy como el primer día, se supone que estarán allí el séptimo y el octavo día. Están pensando en comprar un seguro de viaje que les reembolsará el costo total del paquete vacacional de $\text{\$2500}$ si llueve ambos días, y no se les reembolsará nada en caso contrario. El seguro cuesta $\text{\$200}$. Supongamos que el clima en la ciudad de Centerville cambia según el modelo de CMTD de tres estados (soleado, nublado y lluvioso) visto en clase. Suponiendo que hoy hace sol en la ciudad de Centerville, ¿deberían el Sr. y la Sra. H comprar el seguro?


Cadena de Markov de tiempo discreto visto en clase
$$ P =  \begin{pmatrix}
0.5 & 0.3 & 0.2 \\
0.5 & 0.2 & 0.3 \\
0.4 & 0.5 & 0.1
\end{pmatrix}
$$

\n
$$Estado soleado = 0$$
$$Estado nublado = 1$$
$$Estado nublado = 2$$



```python
import numpy as np
```

```python
# Matriz de transición de 3 estados (soleado, nublado, lluvioso)
P = np.array([
    [0.5, 0.3, 0.2],
    [0.5, 0.2, 0.3],
    [0.4, 0.5, 0.1]
])

# Índices de los estados
estado_soleado = 0
estado_lluvioso = 2

# Días: hoy es día 1 (soleado), viaje en días 7 y 8 → 6 pasos hasta día 7
pasos = 6

# Costos del seguro
costo_seguro = 200
reembolso = 2500
```

```python
# Probabilidad de lluvia en el día 7
P_n = np.linalg.matrix_power(P, pasos) #Calcula la matriz de transición en 6 pasos
                                        #(P⁶). El elemento (i, j) de esta matriz es la probabilidad de estar en el estado j
                                        #después de 6 días, habiendo empezado en el estado i.

prob_lluvia_dia7 = P_n[estado_soleado, estado_lluvioso] #  Extrae la probabilidad de que el día 7 sea lluvioso,
                                                        #dado que hoy (día 1) es soleado. Esto es P⁶[0, 2].

# Probabilidad de lluvia en día 8 dado que día 7 fue lluvioso
prob_lluvia_dia8_dado_lluvia = P[estado_lluvioso, estado_lluvioso] #Probabilidad de que el día 8 sea lluvioso dado que el día 7 fue
                                                                   #lluvioso. Por la propiedad de Markov, es simplemente la entrada de la
                                                                   #matriz de un paso: P[2, 2].

# Probabilidad conjunta
prob_ambos = prob_lluvia_dia7 * prob_lluvia_dia8_dado_lluvia #Probabilidad conjunta de que llueva ambos días.
                                                             #Se multiplican porque la probabilidad del día 8
                                                             #depende solo del día 7 (cadena de Markov).

# Valor esperado del seguro
valor_esperado = reembolso * prob_ambos - costo_seguro #Calcula el beneficio esperado de comprar el seguro:
                                                       #ingreso esperado (reembolso × probabilidad) menos el costo.
                                                       #Si es positivo, conviene comprarlo; si es negativo, no.

# Resultados
print(f"Probabilidad de lluvia en ambos días: {prob_ambos:.6f}")
print(f"Valor esperado del seguro: ${valor_esperado:.2f}")

if valor_esperado > 0:
    print("Conviene comprar el seguro.")
else:
    print("NO conviene comprar el seguro.")
```
